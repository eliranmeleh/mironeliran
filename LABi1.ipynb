```python
# === Mount Google Drive ===
from google.colab import drive
drive.mount('/content/drive')

# === Import Libraries ===
import ipywidgets as widgets
from IPython.display import display, clear_output
import os

# === File Path ===
# Make sure the file students.txt exists in your MyDrive
file_path = '/content/drive/My Drive/students.txt'

# === Step 1: Read Students File ===
if not os.path.exists(file_path):
    raise FileNotFoundError("⚠️ File not found! Please upload students.txt to your Google Drive.")

students = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        parts = [p.strip() for p in line.strip().split(',')]
        if len(parts) < 6:
            continue  # skip incomplete lines
        first, last, email = parts[0], parts[1], parts[2]
        courses = parts[3:-2]
        semester = parts[-2]
        linkedin = parts[-1]
        students.append({
            "first": first,
            "last": last,
            "email": email,
            "courses": courses,
            "semester": semester,
            "linkedin": linkedin
        })

# === Step 2: Create Dropdown Widget ===
student_names = [f"{s['first']} {s['last']}" for s in students]

dropdown = widgets.Dropdown(
    options=student_names,
    description='Select student:',
    style={'description_width': 'initial'}
)

output = widgets.Output()

# Function to display student info
def show_student_info(change):
    output.clear_output()
    with output:
        index = student_names.index(change['new'])
        s = students[index]
        print(f"Name: {s['first']} {s['last']}")
        print(f"Email: {s['email']}")
        print(f"Semester: {s['semester']}")
        print("Courses:")
        for c in s['courses']:
            print(f"  • {c}")
        print(f"LinkedIn: {s['linkedin']}")

# Link dropdown to function
dropdown.observe(show_student_info, names='value')

# Display dropdown and output
display(dropdown, output)

# Show first student automatically
if students:
    show_student_info({'new': student_names[0]})

# === Step 3: Add Text Box and Button for Update ===
text_new_course = widgets.Text(
    description='New course:',
    placeholder='Type the course name here...'
)

button_update = widgets.Button(
    description='Update Student (Add Row)',
    button_style='success'
)

output_msg = widgets.Output()

# Append new line (add, not replace)
def append_student_with_new_course(student, new_course, path=file_path):
    new_courses = student["courses"].copy()
    if new_course and new_course not in new_courses:
        new_courses.append(new_course)
    line = ", ".join([
        student["first"],
        student["last"],
        student["email"],
        *new_courses,
        str(student["semester"]),
        student["linkedin"]
    ]) + "\n"
    with open(path, "a", encoding="utf-8") as f:
        f.write(line)

# Handle button click
def on_update_click(b):
    with output_msg:
        clear_output()
        if not dropdown.value:
            print("⚠️ Please select a student first.")
            return
        new_course = text_new_course.value.strip()
        if not new_course:
            print("⚠️ Please type a course name.")
            return
        index = student_names.index(dropdown.value)
        s = students[index]
        append_student_with_new_course(s, new_course)
        print(f"✅ Added course '{new_course}' for {s['first']} {s['last']}.")
        print("(A new line has been added to students.txt)")

button_update.on_click(on_update_click)

# Display input and button
display(text_new_course, button_update, output_msg)

# === Styling ===
button_update.style.button_color = 'lightgreen'
dropdown.layout.width = '400px'
text_new_course.layout.width = '400px'

import json, copy, os

# === paths ===
SRC = "/content/drive/My Drive/lab1.ipynb"      # << עדכן נתיב למקור
DST = "/content/drive/My Drive/lab1_clean.ipynb"  # << עדכן נתיב לקובץ נקי

def strip_widget_outputs(outputs):
    """Remove widget-view outputs and keep only safe outputs."""
    safe = []
    for out in outputs or []:
        # skip any widget-view payloads
        data = out.get("data", {})
        if "application/vnd.jupyter.widget-view+json" in data:
            continue
        safe.append(out)
    return safe

with open(SRC, "r", encoding="utf-8") as f:
    nb = json.load(f)

nb = copy.deepcopy(nb)

# 1) remove top-level widgets metadata
md = nb.get("metadata", {})
if "widgets" in md:
    md.pop("widgets", None)
nb["metadata"] = md

# 2) normalize nbformat (optional but helps GitHub)
nb["nbformat"] = 4
nb["nbformat_minor"] = 5

# 3) clean every cell: remove referenced_widgets, widget outputs, execution_count
for cell in nb.get("cells", []):
    # metadata.referenced_widgets
    meta = cell.get("metadata", {})
    colab_meta = meta.get("colab", {})
    if "referenced_widgets" in colab_meta:
        colab_meta.pop("referenced_widgets", None)
        meta["colab"] = colab_meta
        cell["metadata"] = meta

    # outputs: drop widget views
    if cell.get("cell_type") == "code":
        cell["outputs"] = strip_widget_outputs(cell.get("outputs", []))
        # אפשר גם לנקות לגמרי אם תרצה:
        # cell["outputs"] = []
        cell["execution_count"] = None

# 4) כתיבה
with open(DST, "w", encoding="utf-8") as f:
    json.dump(nb, f, ensure_ascii=False, indent=2)

print("✅ Wrote cleaned notebook:", DST)

```
